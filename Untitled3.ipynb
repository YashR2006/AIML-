```python

# =============================================================================
# AUTISM SPECTRUM DISORDER (ASD)
# MACHINE LEARNING PROJECT: CLASSIFICATION AND REGRESSION
# DATA: 3,000 automatically generated synthetic records
# CLASSIFICATION: 5 algorithms
# REGRESSION: 3 algorithms
# VISUALIZATION: 14 high-resolution graphs (300 DPI)
# OUTPUT: CSV, Excel, PNG, PKL and ZIP
#
# IMPORTANT:
# Synthetic educational data only.
# NOT real clinical data and NOT suitable for diagnosis.
# =============================================================================


# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import sys
import warnings
import zipfile
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor
from sklearn.ensemble import (
    RandomForestClassifier,
    RandomForestRegressor
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    roc_auc_score,
    roc_curve,
    classification_report
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

print("=" * 85)
print("AUTISM SPECTRUM DISORDER - MACHINE LEARNING PROJECT")
print("CLASSIFICATION AND REGRESSION")
print("=" * 85)


# =============================================================================
# 2. CREATE OUTPUT DIRECTORIES
# =============================================================================

OUTPUT_DIR = Path("ASD_ML_Project_Results")
GRAPH_DIR = OUTPUT_DIR / "Graphs"
MODEL_DIR = OUTPUT_DIR / "Models"
DATA_DIR = OUTPUT_DIR / "Dataset"
REPORT_DIR = OUTPUT_DIR / "Reports"

for folder in [OUTPUT_DIR, GRAPH_DIR, MODEL_DIR, DATA_DIR, REPORT_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

print("\nOutput folders created successfully.")


# =============================================================================
# 3. GENERATE SYNTHETIC ASD-RELATED DATASET
# =============================================================================

print("\nGenerating 3,000 synthetic records...")

N = 3000

rng = np.random.default_rng(RANDOM_STATE)

age = rng.integers(4, 12, N).astype(float)

gender = rng.choice(
    ["Male", "Female"],
    N,
    p=[0.70, 0.30]
)

ethnicity = rng.choice(
    ["White-European", "Asian", "Middle Eastern",
     "Black", "South Asian", "Other"],
    N
)

jaundice = rng.choice(
    ["Yes", "No"],
    N,
    p=[0.25, 0.75]
)

family_history = rng.choice(
    ["Yes", "No"],
    N,
    p=[0.20, 0.80]
)

country = rng.choice(
    ["United Kingdom", "United States", "India",
     "Australia", "Canada", "Other"],
    N
)

previous_app_use = rng.choice(
    ["Yes", "No"],
    N,
    p=[0.10, 0.90]
)

relation = rng.choice(
    ["Parent", "Self", "Relative", "Health Care Professional"],
    N
)

# Generate ten behavioural screening scores.
# These are simulated questionnaire responses, not validated measurements.

score_matrix = rng.binomial(
    1,
    0.55,
    size=(N, 10)
)

score_columns = [f"A{i}_Score" for i in range(1, 11)]

df = pd.DataFrame(score_matrix, columns=score_columns)

df["age"] = age
df["gender"] = gender
df["ethnicity"] = ethnicity
df["jaundice"] = jaundice
df["family_history"] = family_history
df["country"] = country
df["previous_app_use"] = previous_app_use
df["relation"] = relation

# Simulated screening outcome.
# It is generated from questionnaire scores and selected synthetic factors.

total_score = score_matrix.sum(axis=1)

risk_score = (
    0.85 * total_score
    + 0.65 * (family_history == "Yes").astype(int)
    + 0.30 * (jaundice == "Yes").astype(int)
    + rng.normal(0, 1.8, N)
)

probability = 1 / (
    1 + np.exp(-(risk_score - 5.2))
)

asd_class = rng.binomial(1, probability)

df["Class_ASD"] = np.where(
    asd_class == 1,
    "YES",
    "NO"
)

# Add small amounts of missing data to demonstrate preprocessing.

for column in ["age", "ethnicity", "relation"]:
    missing_indices = rng.choice(
        N,
        size=int(N * 0.025),
        replace=False
    )
    df.loc[missing_indices, column] = np.nan

print("Dataset generated successfully!")
print("Dataset shape:", df.shape)


# =============================================================================
# 4. DATASET OVERVIEW
# =============================================================================

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nFirst 10 records:")
print(df.head(10).to_string())

print("\nDataset dimensions:")
print(df.shape)

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nStatistical summary:")
print(df.describe(include="all").to_string())

print("\nClass distribution:")
print(df["Class_ASD"].value_counts())

df.to_csv(
    DATA_DIR / "synthetic_asd_dataset.csv",
    index=False
)


# =============================================================================
# 5. EXPLORATORY DATA ANALYSIS
# =============================================================================

print("\nPerforming exploratory data analysis...")

sns.set_theme(style="whitegrid")

# GRAPH 1: CLASS DISTRIBUTION

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Class_ASD"
)

plt.title("ASD Screening Class Distribution")
plt.xlabel("Screening Class")
plt.ylabel("Number of Records")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "01_class_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 2: AGE DISTRIBUTION

plt.figure(figsize=(9, 5))

sns.histplot(
    df["age"].dropna(),
    bins=15,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age (Years)")
plt.ylabel("Frequency")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "02_age_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 3: QUESTIONNAIRE SCORE DISTRIBUTION

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=df[score_columns]
)

plt.title("Distribution of Questionnaire Scores")
plt.xlabel("Questionnaire")
plt.ylabel("Score")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "03_questionnaire_scores.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 4: CORRELATION HEATMAP

numeric_df = df[score_columns + ["age"]].copy()

numeric_df["Class_ASD_Numeric"] = (
    df["Class_ASD"].map({"NO": 0, "YES": 1})
)

correlation = numeric_df.corr()

plt.figure(figsize=(13, 9))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of ASD Dataset")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "04_correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 5: MISSING VALUE ANALYSIS

plt.figure(figsize=(10, 5))

missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > 0]

if len(missing_counts) > 0:
    missing_counts.sort_values().plot(
        kind="bar"
    )
    plt.title("Missing Values by Feature")
    plt.xlabel("Feature")
    plt.ylabel("Missing Count")
else:
    plt.text(0.5, 0.5, "No missing values", ha="center")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "05_missing_values.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 6. DATA PREPROCESSING
# =============================================================================

print("\n" + "=" * 70)
print("DATA PREPROCESSING")
print("=" * 70)

# Convert target to numerical format.

df["Class_ASD"] = df["Class_ASD"].map({
    "NO": 0,
    "YES": 1
})

numeric_features = score_columns + ["age"]

categorical_features = [
    "gender",
    "ethnicity",
    "jaundice",
    "family_history",
    "country",
    "previous_app_use",
    "relation"
]

# Compatible with current and older scikit-learn versions.

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        drop="first",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        drop="first",
        sparse=False
    )

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(
            strategy="constant",
            fill_value="Unknown"
        )
    ),
    ("encoder", encoder)
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

X = df[numeric_features + categorical_features].copy()
y = df["Class_ASD"].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("\nTraining records:", len(X_train))
print("Testing records:", len(X_test))

print("\nFitting preprocessing pipeline...")

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing completed.")
print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

# Save preprocessing object.

joblib.dump(
    preprocessor,
    MODEL_DIR / "classification_preprocessor.pkl"
)


# =============================================================================
# 7. CLASSIFICATION MODELS
# =============================================================================

print("\n" + "=" * 70)
print("CLASSIFICATION MODEL TRAINING")
print("=" * 70)

classification_models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    ),

    "Naive Bayes": GaussianNB(),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        max_depth=12,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}

classification_results = []
classification_predictions = {}
classification_probabilities = {}
trained_classification_models = {}

for name, model in classification_models.items():

    print("\nTraining:", name)

    model.fit(
        X_train_processed,
        y_train
    )

    predictions = model.predict(
        X_test_processed
    )

    probabilities = model.predict_proba(
        X_test_processed
    )[:, 1]

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    auc = roc_auc_score(
        y_test,
        probabilities
    )

    classification_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1_Score": f1,
        "ROC_AUC": auc
    })

    classification_predictions[name] = predictions
    classification_probabilities[name] = probabilities
    trained_classification_models[name] = model

    joblib.dump(
        model,
        MODEL_DIR / (
            name.lower().replace(" ", "_") + "_classifier.pkl"
        )
    )

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))
    print("ROC AUC  :", round(auc, 4))


classification_results_df = pd.DataFrame(
    classification_results
)

classification_results_df = (
    classification_results_df.sort_values(
        by="F1_Score",
        ascending=False
    )
)

print("\nCLASSIFICATION COMPARISON")
print(classification_results_df.to_string(index=False))

best_classifier_name = classification_results_df.iloc[0]["Model"]

best_classifier = trained_classification_models[
    best_classifier_name
]

best_predictions = classification_predictions[
    best_classifier_name
]

print("\nBest classifier:", best_classifier_name)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        best_predictions,
        zero_division=0
    )
)


# =============================================================================
# 8. CLASSIFICATION VISUALIZATIONS
# =============================================================================

# GRAPH 6: CONFUSION MATRIX

cm = confusion_matrix(
    y_test,
    best_predictions
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["NO", "YES"],
    yticklabels=["NO", "YES"]
)

plt.title(
    f"Confusion Matrix - {best_classifier_name}"
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "06_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 7: MODEL ACCURACY COMPARISON

plt.figure(figsize=(11, 6))

sns.barplot(
    data=classification_results_df,
    x="Model",
    y="Accuracy"
)

plt.title("Classification Model Accuracy Comparison")
plt.xlabel("Machine Learning Algorithm")
plt.ylabel("Accuracy")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "07_classification_accuracy.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 8: CLASSIFICATION METRICS COMPARISON

metrics_long = classification_results_df.melt(
    id_vars="Model",
    value_vars=[
        "Accuracy",
        "Precision",
        "Recall",
        "F1_Score"
    ],
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=metrics_long,
    x="Model",
    y="Score",
    hue="Metric"
)

plt.title("Classification Evaluation Metrics")
plt.ylabel("Score")
plt.xlabel("Classification Algorithm")
plt.xticks(rotation=25, ha="right")
plt.legend(title="Evaluation Metric")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "08_classification_metrics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 9: ROC CURVE

plt.figure(figsize=(9, 7))

for name in classification_models:

    probabilities = classification_probabilities[name]

    fpr, tpr, _ = roc_curve(
        y_test,
        probabilities
    )

    auc_value = roc_auc_score(
        y_test,
        probabilities
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{name} (AUC={auc_value:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("ROC Curve Comparison")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "09_roc_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 10: PRECISION-RECALL CURVE

from sklearn.metrics import precision_recall_curve

plt.figure(figsize=(9, 7))

for name in classification_models:

    precision_values, recall_values, _ = (
        precision_recall_curve(
            y_test,
            classification_probabilities[name]
        )
    )

    plt.plot(
        recall_values,
        precision_values,
        label=name
    )

plt.title("Precision-Recall Curve Comparison")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "10_precision_recall_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 9. REGRESSION TASK
# =============================================================================

print("\n" + "=" * 70)
print("REGRESSION MODEL TRAINING")
print("=" * 70)

# Regression target: AGE.
# Age is predicted from questionnaire and demographic features.
# Age is excluded from regression input to prevent target leakage.

regression_features = (
    score_columns + categorical_features
)

X_reg = df[regression_features].copy()
y_reg = df["age"].copy()

# Exclude rows with missing regression targets.

valid_rows = y_reg.notna()

X_reg = X_reg.loc[valid_rows]
y_reg = y_reg.loc[valid_rows]

X_reg_train, X_reg_test, y_reg_train, y_reg_test = (
    train_test_split(
        X_reg,
        y_reg,
        test_size=0.20,
        random_state=RANDOM_STATE
    )
)

regression_preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, score_columns),
    ("categorical", categorical_pipeline, categorical_features)
])

X_reg_train_processed = (
    regression_preprocessor.fit_transform(
        X_reg_train
    )
)

X_reg_test_processed = (
    regression_preprocessor.transform(
        X_reg_test
    )
)

joblib.dump(
    regression_preprocessor,
    MODEL_DIR / "regression_preprocessor.pkl"
)

regression_models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree Regressor": DecisionTreeRegressor(
        max_depth=8,
        random_state=RANDOM_STATE
    ),

    "Random Forest Regressor": RandomForestRegressor(
        n_estimators=200,
        max_depth=12,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}

regression_results = []
regression_predictions = {}
trained_regression_models = {}

for name, model in regression_models.items():

    print("\nTraining:", name)

    model.fit(
        X_reg_train_processed,
        y_reg_train
    )

    predictions = model.predict(
        X_reg_test_processed
    )

    mae = mean_absolute_error(
        y_reg_test,
        predictions
    )

    mse = mean_squared_error(
        y_reg_test,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_reg_test,
        predictions
    )

    regression_results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2_Score": r2
    })

    regression_predictions[name] = predictions
    trained_regression_models[name] = model

    joblib.dump(
        model,
        MODEL_DIR / (
            name.lower().replace(" ", "_") + "_regressor.pkl"
        )
    )

    print("MAE :", round(mae, 4))
    print("MSE :", round(mse, 4))
    print("RMSE:", round(rmse, 4))
    print("R2  :", round(r2, 4))


regression_results_df = pd.DataFrame(
    regression_results
)

regression_results_df = (
    regression_results_df.sort_values(
        by="RMSE",
        ascending=True
    )
)

print("\nREGRESSION COMPARISON")
print(regression_results_df.to_string(index=False))

best_regressor_name = regression_results_df.iloc[0]["Model"]

best_regressor = trained_regression_models[
    best_regressor_name
]

best_regression_predictions = regression_predictions[
    best_regressor_name
]

print("\nBest regressor:", best_regressor_name)


# =============================================================================
# 10. REGRESSION VISUALIZATIONS
# =============================================================================

# GRAPH 11: ACTUAL VS PREDICTED SCATTER PLOT

plt.figure(figsize=(9, 7))

plt.scatter(
    y_reg_test,
    best_regression_predictions,
    alpha=0.65,
    label="Predicted values"
)

min_value = min(
    y_reg_test.min(),
    best_regression_predictions.min()
)

max_value = max(
    y_reg_test.max(),
    best_regression_predictions.max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--",
    label="Ideal prediction"
)

# Regression trend line

slope, intercept = np.polyfit(
    y_reg_test,
    best_regression_predictions,
    1
)

trend_x = np.linspace(
    min_value,
    max_value,
    100
)

plt.plot(
    trend_x,
    slope * trend_x + intercept,
    label="Regression trend line"
)

plt.title(
    f"Actual vs Predicted Age - {best_regressor_name}"
)

plt.xlabel("Actual Age")
plt.ylabel("Predicted Age")
plt.legend()
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "11_actual_vs_predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 12: RESIDUAL PLOT

residuals = (
    y_reg_test.values -
    best_regression_predictions
)

plt.figure(figsize=(9, 6))

plt.scatter(
    best_regression_predictions,
    residuals,
    alpha=0.65
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Regression Residual Plot")
plt.xlabel("Predicted Age")
plt.ylabel("Residual Error")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "12_regression_residuals.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 13: REGRESSION MODEL COMPARISON

plt.figure(figsize=(10, 6))

sns.barplot(
    data=regression_results_df,
    x="Model",
    y="RMSE"
)

plt.title("Regression Model RMSE Comparison")
plt.xlabel("Regression Algorithm")
plt.ylabel("RMSE")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "13_regression_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# GRAPH 14: REGRESSION METRICS

regression_metrics_long = regression_results_df.melt(
    id_vars="Model",
    value_vars=["MAE", "MSE", "RMSE", "R2_Score"],
    var_name="Metric",
    value_name="Value"
)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=regression_metrics_long,
    x="Model",
    y="Value",
    hue="Metric"
)

plt.title("Regression Evaluation Metrics")
plt.xlabel("Regression Algorithm")
plt.ylabel("Metric Value")
plt.xticks(rotation=20, ha="right")
plt.legend(title="Metric")
plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "14_regression_metrics.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# =============================================================================
# 11. SAVE RESULTS
# =============================================================================

print("\n" + "=" * 70)
print("SAVING PROJECT RESULTS")
print("=" * 70)

classification_results_df.to_csv(
    OUTPUT_DIR / "classification_results.csv",
    index=False
)

regression_results_df.to_csv(
    OUTPUT_DIR / "regression_results.csv",
    index=False
)

# Save all result tables in one Excel workbook.

with pd.ExcelWriter(
    OUTPUT_DIR / "ASD_ML_Results.xlsx"
) as writer:

    classification_results_df.to_excel(
        writer,
        sheet_name="Classification",
        index=False
    )

    regression_results_df.to_excel(
        writer,
        sheet_name="Regression",
        index=False
    )

    df.to_excel(
        writer,
        sheet_name="Synthetic Dataset",
        index=False
    )

# Save actual and predicted regression values.

regression_prediction_df = pd.DataFrame({
    "Actual_Age": y_reg_test.values,
    "Predicted_Age": best_regression_predictions,
    "Residual": residuals
})

regression_prediction_df.to_csv(
    OUTPUT_DIR / "regression_predictions.csv",
    index=False
)

# Save classification predictions.

classification_prediction_df = pd.DataFrame({
    "Actual_Class": y_test.values
})

for name, predictions in classification_predictions.items():
    classification_prediction_df[
        name
    ] = predictions

classification_prediction_df.to_csv(
    OUTPUT_DIR / "classification_predictions.csv",
    index=False
)

# Save text report.

with open(
    REPORT_DIR / "model_summary.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write("ASD MACHINE LEARNING PROJECT\n")
    file.write("=" * 60 + "\n")

    file.write(
        "\nDATA SOURCE: Synthetic educational dataset\n"
    )

    file.write(
        f"\nNumber of records: {N}\n"
    )

    file.write(
        f"\nNumber of features: {len(df.columns) - 1}\n"
    )

    file.write(
        "\nBEST CLASSIFIER: " + best_classifier_name + "\n"
    )

    file.write(
        "\nCLASSIFICATION RESULTS:\n"
    )

    file.write(
        classification_results_df.to_string(index=False)
    )

    file.write(
        "\n\nBEST REGRESSOR: " + best_regressor_name + "\n"
    )

    file.write(
        "\nREGRESSION RESULTS:\n"
    )

    file.write(
        regression_results_df.to_string(index=False)
    )

    file.write(
        "\n\nIMPORTANT: Synthetic data is not clinical evidence.\n"
    )


# =============================================================================
# 12. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 85)
print("FINAL CLASSIFICATION RESULTS")
print("=" * 85)

print(
    classification_results_df.to_string(index=False)
)

print("\nBest classifier:", best_classifier_name)

print("\n" + "=" * 85)
print("FINAL REGRESSION RESULTS")
print("=" * 85)

print(
    regression_results_df.to_string(index=False)
)

print("\nBest regressor:", best_regressor_name)


# =============================================================================
# 13. CREATE FINAL ZIP FILE
# =============================================================================

print("\nCreating final project ZIP file...")

zip_path = Path("ASD_ML_Complete_Project.zip")

with zipfile.ZipFile(
    zip_path,
    "w",
    zipfile.ZIP_DEFLATED
) as zip_file:

    for root, dirs, files in os.walk(OUTPUT_DIR):

        for filename in files:

            file_path = Path(root) / filename

            zip_file.write(
                file_path,
                file_path.relative_to(OUTPUT_DIR.parent)
            )

print("\nZIP created successfully:", zip_path)


# =============================================================================
# 14. PROJECT COMPLETION
# =============================================================================

print("\n" + "=" * 85)
print("ASD MACHINE LEARNING PROJECT COMPLETED SUCCESSFULLY")
print("=" * 85)

print("\nPROJECT OUTPUTS:")
print("1. Synthetic ASD Dataset")
print("2. Five Classification Models")
print("3. Three Regression Models")
print("4. Classification Evaluation Metrics")
print("5. Regression Evaluation Metrics")
print("6. Fourteen High-Resolution Graphs")
print("7. Confusion Matrix")
print("8. Correlation Heatmap")
print("9. ROC Curve")
print("10. Precision-Recall Curve")
print("11. Actual vs Predicted Scatter Plot")
print("12. Residual Analysis")
print("13. Excel Results Workbook")
print("14. CSV Result Files")
print("15. Saved ML Models")
print("16. Final ZIP Package")

print("\nOutput directory:", OUTPUT_DIR.resolve())
print("ZIP file:", zip_path.resolve())

print("\nAll graphs have been saved at 300 DPI.")

print("\nIMPORTANT:")
print("This project uses synthetic educational data.")
print("Its performance metrics do not establish real-world")
print("ASD screening accuracy or diagnostic validity.")

# Download ZIP automatically in Google Colab.

try:
    from google.colab import files
    files.download(str(zip_path))
except ImportError:
    print("\nNot running in Google Colab.")
    print("ZIP file is available in the current working directory.")
```
